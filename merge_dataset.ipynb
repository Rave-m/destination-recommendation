```python
import pandas as pd
import numpy as np
```

```python
reviews = 'results/reviews.csv'
details = 'results/details.csv'

reviews_df = pd.read_csv(reviews)
details_df = pd.read_csv(details)
```

```python
reviews_df.head()
```

```python
details_df.head()
```

```python
reviews_df.info()
```

```python
details_df.info()
```

```python
destinasi_review = reviews_df['nama_destinasi'].unique().tolist()
destinasi_detail = details_df['nama_destinasi'].unique().tolist()

print(f"destinasi reviews {len(destinasi_review)} : ")
print(f"{destinasi_review}")

print("\n")
print(f"destinasi details {len(destinasi_detail)}")
print(f"{destinasi_detail}")
```

```python
tidak_ada_di_reviews = list(set(destinasi_detail) - set(destinasi_review))
print(f"🔵 Destinasi yang ADA DI DETAILS tapi TIDAK DI REVIEWS ({len(tidak_ada_di_reviews)}):")
for i, d in enumerate(tidak_ada_di_reviews, 1):
    print(f"{i}. {d}")
```

```python
destinations_df = pd.merge(
    reviews_df,
    details_df,
    on='nama_destinasi',
    how='inner'
)

destinations_df.head()
```

```python
destinations_df.info()
```

## **HANDLING NULL**

```python
destinations_df.isna().sum()
```

```python
destinations_df[destinations_df['username'].isnull()]
```

```python
# drop kolom username
destinations_df.drop(columns="username", inplace=True)
```

```python
destinations_df.info()
```

```python
# shuffle dataset
destinations_df = destinations_df.sample(frac=1).reset_index(drop=True)
```

```python
destinations_df.head()
```

```python
# simpan ke csv
destinations_df.to_csv("dataset/destinations.csv", index=False)
```
